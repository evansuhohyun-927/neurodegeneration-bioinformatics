### 예시 3: 떨림 주파수를 숫자로 느껴 보기

"5 Hz로 떨린다"는 말은 1초에 5번 왕복한다는 뜻입니다. 가상의 신호를 만들어 실제로 세어 봅시다. (내일 자세히 배우니, 오늘은 붙여 넣고 결과만 보세요.)

```python
import numpy as np

fs = 100                      # 1초에 100번 측정 (100 Hz)
t = np.arange(0, 10, 1 / fs)  # 10초 동안의 시간축
tremor = 0.05 * np.sin(2 * np.pi * 5 * t)   # 5 Hz로 떨리는 가상 손목 신호

# 신호가 음(-)에서 양(+)으로 넘어가는 횟수 = 진동 횟수
crossings = np.sum((tremor[:-1] < 0) & (tremor[1:] >= 0))
print("측정점 개수:", len(t))
print("10초 동안 진동 횟수:", crossings)
print("추정 주파수(Hz):", crossings / 10)
```
