### 6. 대사 비용: rate와 cost

아래 코드는 이 노트를 위해 새로 작성했습니다. 두 식과 역진자의 에너지 교환을 함께 계산합니다.

```python
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

# 1) Whittle 6판 2장이 소개하는 두 식 (v: m/s)
v = np.linspace(0.5, 2.0, 301)
Ew = 2.23 + 1.26*v**2            # 단위 시간당 (W/kg)
Em = 2.23/v + 1.26*v             # 단위 거리당 (J/(m·kg))
v_opt = np.sqrt(2.23/1.26)
print(f"Em 최소 속도 = {v_opt:.3f} m/s, 그때 Em = {2.23/v_opt + 1.26*v_opt:.3f} J/(m·kg)")
print(f"70 kg, 1.4 m/s: {70*(2.23+1.26*1.4**2):.0f} W / 70 kg, 최소점: {70*(2.23/v_opt+1.26*v_opt):.0f} J/m")

# 2) Perry 24장(Waters)의 성인 선형식 (V: m/min)
for V in (30, 60, 80, 100):
    rate = 0.129*V + 2.60        # mL O2/(kg·min)
    print(f"V={V:3d} m/min → O2 rate {rate:5.2f} mL/kg/min, O2 cost {rate/V:.3f} mL/kg/m")

# 3) 강체 다리 역진자 한 걸음: 위치에너지와 운동에너지가 맞바뀐다
g, L, m = 9.81, 0.9, 70.0
v_top = 1.1                                   # 다리가 수직일 때 COM 속도 (m/s), 가정
th = np.radians(np.linspace(-20, 20, 81))     # 다리 각 (수직 = 0)
h = L*np.cos(th)
v2 = v_top**2 + 2*g*(L - h)                   # 에너지 보존
PE, KE = m*g*h, 0.5*m*v2
print(f"PE 범위 {PE.max()-PE.min():.1f} J, KE 범위 {KE.max()-KE.min():.1f} J, PE+KE 변동 {np.ptp(PE+KE):.2e} J")

fig, ax = plt.subplots(1, 2, figsize=(9, 3.2))
ax[0].plot(v, Em); ax[0].axvline(v_opt, ls="--", c="gray")
ax[0].set_xlabel("speed (m/s)"); ax[0].set_ylabel("J/(m·kg)"); ax[0].set_title("cost per distance")
ax[1].plot(np.degrees(th), PE - PE.min(), label="PE - min")
ax[1].plot(np.degrees(th), KE - KE.min(), label="KE - min")
ax[1].set_xlabel("leg angle (deg)"); ax[1].set_ylabel("J"); ax[1].legend(); ax[1].set_title("inverted pendulum")
fig.tight_layout(); fig.savefig("T05_energy.png", dpi=120)
print("그림 저장: T05_energy.png")
```
