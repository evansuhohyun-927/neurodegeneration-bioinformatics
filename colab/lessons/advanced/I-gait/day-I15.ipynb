## 예시로 이해하기

### 예시 1 — Trendelenburg 보행을 숫자로: 몸통을 기울이면 외전근이 얼마나 편해지나

앞에서 본 그림(관상면)에서 지지 고관절을 받침점으로 놓으면, 고관절 안쪽에 얹힌 체중의 모멘트를 외전근이 반대 방향 모멘트로 버텨야 골반이 수평으로 유지됩니다. 몸통을 지지 쪽으로 기울이면 무게중심이 받침점에 가까워져 체중의 지렛대 팔이 짧아집니다. 아래 숫자는 **설명용 가정값**이며 교재의 예제가 아닙니다.

```python
# 한 발 서기에서 고관절 외전근이 버텨야 하는 힘: 시상면 대신 관상면(앞에서 본 그림)의 지렛대 평형
# 숫자는 모두 설명용 가정값이다(교재 예제 아님). 수직 방향 힘만 보는 단순화.
g = 9.81
body_mass = 70.0                     # kg
leg_frac = 0.16                      # 한쪽 다리 질량 ≈ 체중의 16% (인체계측 표의 대략값, I4 참고)
W = body_mass * (1 - leg_frac) * g   # 지지 고관절 위에 얹힌 무게(들린 다리 포함, 지지 다리 제외), N
d_abd = 0.05                         # 외전근의 지렛대 팔(고관절 중심 → 대전자 쪽 근육 작용선), m

for label, d_com in [("몸통 곧게(무게중심이 고관절 안쪽 10 cm)", 0.10),
                     ("지지 쪽으로 몸통 기울임(4 cm)", 0.04),
                     ("크게 기울임(1 cm)", 0.01)]:
    F_abd = W * d_com / d_abd                          # 모멘트 평형: F_abd × d_abd = W × d_com
    F_joint = F_abd + W                                # 고관절이 받는 힘(같은 방향으로 단순화)
    print(f"{label:34s} 외전근 {F_abd:6.0f} N ({F_abd / (body_mass * g):.1f} BW) | "
          f"고관절 {F_joint:6.0f} N ({F_joint / (body_mass * g):.1f} BW)")
```
