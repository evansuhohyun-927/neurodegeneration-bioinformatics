## 예시로 이해하기

### 예시 1 — Daphnet의 FOG 삽화는 얼마나 짧은가

내일 쓸 Daphnet 데이터(설명과 내려받기는 [Day I18](day-I18.md))의 라벨만 읽어 삽화 길이 분포를 봅니다. 4초 window로 FI를 계산하기 전에, 4초보다 짧은 삽화가 얼마나 되는지 알아 두면 결과 해석이 달라집니다.

```python
import glob, os
import numpy as np

DATA = "data/daphnet/dataset_fog_release/dataset"
fs = 64
dur = []
for path in sorted(glob.glob(os.path.join(DATA, "S*R*.txt"))):
    lab = np.loadtxt(path, usecols=10)
    fog = (lab == 2).astype(int)
    edges = np.diff(np.r_[0, fog, 0])                 # +1 = 시작, −1 = 끝
    starts, stops = np.where(edges == 1)[0], np.where(edges == -1)[0]
    dur += list((stops - starts) / fs)
dur = np.array(dur)
print(f"FOG 삽화 {len(dur)}개 | 길이 중앙값 {np.median(dur):.1f} s, 평균 {dur.mean():.1f} s, "
      f"범위 {dur.min():.1f}–{dur.max():.1f} s")
for cut in [2, 4, 10, 20]:
    print(f"  {cut:2d} s보다 짧은 삽화: {100 * (dur < cut).mean():.1f}%")
```
