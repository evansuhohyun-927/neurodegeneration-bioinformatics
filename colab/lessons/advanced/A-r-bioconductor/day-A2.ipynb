### 예시 2 — 벡터 연산과 재활용

```r
ctrl  <- c(Gfap = 400, Aif1 = 210, Cd68 = 50, Tmem119 = 300)
treat <- c(Gfap = 900, Aif1 = 400, Cd68 = 210, Tmem119 = 290)
log2fc <- log2((treat + 1) / (ctrl + 1))
round(log2fc, 2)
1:6 + c(0, 100)          # 짧은 쪽이 되풀이된다(recycling)
1:5 + c(0, 100)          # 길이가 배수가 아니면 경고
```

### 예시 3 — 네 가지 열쇠와 `%in%`

```r
lfc <- c(Gfap = 1.17, Aif1 = 0.93, Cd68 = 2.05, Tmem119 = -0.05, Il1b = 1.60)
lfc[c(1, 3)]                 # 양수 위치
lfc[-c(1, 3)]                # 음수 위치 = 빼기
lfc[lfc > 1]                 # 논리 벡터
lfc[c("Cd68", "Gfap")]       # 이름
which(lfc > 1)
microglia <- c("Aif1", "Tmem119", "P2ry12")
microglia %in% names(lfc)
lfc[names(lfc) %in% microglia]
lfc[lfc > 1] <- round(lfc[lfc > 1])   # 부분 대입
lfc
```

### 예시 4 — 프로브 이름 다듬기와 서열 문자열

```r
ids <- c("Gfap_probe001", "Aif1_probe002", "NEG_C_probe003")
nchar(ids)
grepl("^NEG", ids)
sub("_probe[0-9]+$", "", ids)
tolower("GFAP")
sprintf("%s: log2FC = %.2f", "Cd68", 2.0456)
revcomp <- function(s) paste(rev(strsplit(chartr("ACGT", "TGCA", s), "")[[1]]), collapse = "")
revcomp("ATGCCGTA")
```
