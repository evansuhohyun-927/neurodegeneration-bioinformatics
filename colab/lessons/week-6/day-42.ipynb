### 예시 3 — 시간 예산 계산기

```python
plan = [
    ("질문",        40),
    ("데이터",      60),
    ("QC",          50),
    ("결과",       120),
    ("한계",        60),
    ("다음 실험",   40),
]

total = sum(seconds for _, seconds in plan)
for name, seconds in plan:
    bar = "#" * (seconds // 10)
    print(f"{name:<6} {seconds:>4}초 {bar}")
print(f"합계: {total}초 = {total // 60}분 {total % 60}초")
print("5~7분 안에 들어오는가?", 300 <= total <= 420)

# 발표 노트 분량 가늠: 한국어 발표는 대략 1분에 300자 안팎 (사람마다 다름 → 리허설로 확인)
chars_per_min = 300
print("필요한 발표 노트 분량(대략):", round(total / 60 * chars_per_min), "자")
```
