### 예시 2 — 위험 표현 찾기 스크립트

보고서를 눈으로만 읽으면 익숙한 문장은 그냥 지나칩니다. 의심스러운 단어가 들어 있는 줄을 기계적으로 뽑아 주는 작은 도구를 만듭니다.

```python
import re

# 실제로는 open("reports/...md", encoding="utf-8").read() 로 내 보고서를 읽는다
report = """우주비행은 마우스 뇌에 신경염증을 유발한다.
비행군에서 수초 관련 유전자 발현이 낮게 관찰되었다.
이 모델은 파킨슨병을 진단할 수 있다.
이 결과는 우주비행사의 파킨슨병 위험을 증명한다.
PD군의 걸음 간격 변동성이 대조군보다 컸다 (Cohen's d = 1.1).
"""

red_flags = {
    "인과 언어": ["유발", "일으킨", "때문에", "원인", "초래", "증명", "입증"],
    "임상 주장": ["진단할 수", "진단한다", "치료", "예방", "위험을"],
    "과장 부사": ["명백히", "확실히", "분명히", "완벽"],
}

count = 0
for line_no, line in enumerate(report.splitlines(), start=1):
    for category, words in red_flags.items():
        for word in words:
            if re.search(word, line):
                count += 1
                print(f"{line_no}행 [{category}] '{word}' → {line}")
print()
print("다시 읽어볼 곳:", count, "군데")
```
