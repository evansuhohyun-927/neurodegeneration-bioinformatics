### 예시 3 — 가설 그림을 코드로 그리기

실선과 점선을 구분한 그림을 `matplotlib`으로 그립니다. 한글 글꼴은 컴퓨터마다 설정이 달라 깨지기 쉬우므로 그림 안의 글자는 영어로 씁니다.

```python
import os
import matplotlib
matplotlib.use("Agg")          # 창을 띄우지 않고 파일로만 저장
import matplotlib.pyplot as plt

os.makedirs("figures", exist_ok=True)

fig, ax = plt.subplots(figsize=(10, 6))
ax.set_xlim(0, 10)
ax.set_ylim(0, 6.6)
ax.axis("off")

def box(x, y, text, color):
    ax.text(x, y, text, ha="center", va="center", fontsize=10,
            bbox=dict(boxstyle="round,pad=0.5", facecolor=color, edgecolor="black"))

def arrow(x1, y1, x2, y2, style):
    ax.annotate("", xy=(x2, y2), xytext=(x1, y1),
                arrowprops=dict(arrowstyle="->", linestyle=style, color="black", lw=1.5))

# 왼쪽: 내가 관찰한 것 (마우스)
box(1.6, 5.2, "OBSERVED (mouse brain)\nOSD-536: expression changes in\ninflammation / ROS / myelin genes", "#cfe8ff")
# 오른쪽: 내가 관찰한 것 (사람)
box(8.4, 5.2, "OBSERVED (human)\nPhysioNet PD: motor features\ndiffer between PD and controls", "#ffe3c2")

# 가운데: 문헌이 말하는 기전 (내 데이터가 아님)
box(5.0, 3.6, "LITERATURE (not my data)\nneuroinflammation - oxidative stress -\nmitochondria - synapse / myelin", "#e6e6e6")
box(5.0, 1.6, "LITERATURE (not my data)\nbasal ganglia motor circuit function", "#e6e6e6")

# 실선 = 문헌 근거가 있는 연결, 점선 = 가설일 뿐인 연결
arrow(2.3, 4.55, 3.7, 4.0, "dashed")
arrow(5.0, 3.0, 5.0, 2.1, "solid")
arrow(6.3, 2.0, 8.0, 4.5, "solid")

ax.text(5.0, 0.5, "solid = supported by literature in PD     dashed = HYPOTHESIS only (untested here)",
        ha="center", fontsize=9)
ax.text(5.0, 6.3, "Exploratory mechanism hypothesis (no causal claim, datasets not combined)",
        ha="center", fontsize=11, fontweight="bold")

fig.savefig("figures/mechanism_hypothesis.png", dpi=150, bbox_inches="tight")
print("저장 완료: figures/mechanism_hypothesis.png")
```
