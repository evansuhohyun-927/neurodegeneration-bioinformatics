### 예시 1 — 비교 범위 표를 코드로 만들어 보기

표는 손으로 써도 되지만, 코드로 만들면 나중에 항목을 추가하기 쉽습니다. 아래 표의 내용은 실라부스에 적힌 수준의 일반적인 설명입니다. 표본 수·기기 이름 같은 구체적인 값은 여러분이 Day 12와 Day 24에 만든 데이터 사전에서 옮겨 적으세요.

```python
rows = [
    ("종(species)",        "마우스",                    "사람"),
    ("재료/기기",          "뇌 조직 (분자 패널)",       "손목 스마트워치 또는 보행 센서"),
    ("측정 단위",          "유전자별 발현량 (count)",   "가속도·시간 간격 등 운동 신호"),
    ("연구 설계",          "실험 (비행군 vs 지상 대조군)", "관찰 (PD군 vs 대조군)"),
    ("분석 단위",          "마우스 1마리 = 샘플 1개",   "참가자 1명 (반복 측정 포함)"),
    ("주요 outcome",       "log2 fold change, FDR",     "feature의 군 간 효과크기"),
]

lines = ["| 항목 | OSD-536 | PhysioNet PD |", "|---|---|---|"]
for item, osdr, pd_ in rows:
    lines.append(f"| {item} | {osdr} | {pd_} |")
table = "\n".join(lines)
print(table)

same = [item for item, osdr, pd_ in rows if osdr == pd_]
print()
print("두 데이터에서 똑같은 항목 수:", len(same))
```

### 예시 2 — 짝이 없는 데이터를 억지로 짝지으면

서로 아무 관계가 없는 가상의 숫자 두 묶음(마우스 6마리, 사람 6명)을 만들고, "1번 마우스 ↔ 1번 사람" 식으로 순서대로 짝지어 상관계수를 구해 봅니다.

```python
import numpy as np

rng = np.random.default_rng(42)

# 서로 아무 관계 없는 두 숫자 묶음
mouse_gene = rng.normal(loc=10, scale=1, size=6)    # 마우스 6마리의 가상 발현량
human_tremor = rng.normal(loc=3, scale=1, size=6)   # 사람 6명의 가상 떨림 점수

r = np.corrcoef(mouse_gene, human_tremor)[0, 1]
print("마우스 1~6번과 사람 1~6번을 '그냥 순서대로' 짝지은 상관계수:", round(r, 2))

# 짝을 무작위로 바꿔 가며 1000번 반복
rs = []
for _ in range(1000):
    shuffled = rng.permutation(human_tremor)
    rs.append(np.corrcoef(mouse_gene, shuffled)[0, 1])
rs = np.array(rs)
print("짝을 바꿀 때마다 상관계수 범위:", round(rs.min(), 2), "~", round(rs.max(), 2))
print("|r| > 0.7 인 경우의 비율:", round((np.abs(rs) > 0.7).mean(), 3))
```
