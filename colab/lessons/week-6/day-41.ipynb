### 예시 3 — 필수 절이 다 있는지 확인하는 코드

```python
# 실제로는 open("reports/final_report.md", encoding="utf-8").read() 로 읽는다
draft = """# 우주비행 마우스 뇌 분자 변화와 파킨슨병 운동 표현형: 탐색적 분석
## 1. 초록
## 2. 연구 질문과 사전 가설
## 3. 데이터 출처, 버전, 라이선스, 인용
## 4. OSDR 마우스 뇌 분자 데이터의 설계와 QC
## 5. OSDR 분석 결과와 생물학적 해석
## 6. PhysioNet PD 데이터의 설계와 QC
## 7. PD 운동/웨어러블 분석 결과
## 8. 탐색적 기전 연결
## 10. 다음 단계
"""

required = ["초록", "연구 질문", "데이터 출처", "OSDR 마우스 뇌 분자 데이터의 설계와 QC",
            "OSDR 분석 결과", "PhysioNet PD 데이터의 설계와 QC", "PD 운동",
            "탐색적 기전 연결", "민감도 분석과 한계", "다음 단계", "재현 방법"]

headings = [line for line in draft.splitlines() if line.startswith("#")]
missing = [name for name in required if not any(name in h for h in headings)]

print("찾은 제목 수:", len(headings))
print("필수 절 11개 중 빠진 것:", missing)
```
