### 예시 3 — 딕셔너리로 기전 표 만들기

0주차에 배운 딕셔너리와 반복문으로 개념 지도의 표 한 조각을 자동으로 만들 수 있습니다. 아래 유전자들은 마우스 뇌 연구에서 각 기전의 표지로 자주 쓰이는 **예시**입니다. OSD-536의 panel에 실제로 들어 있는지는 2주차 Day 12에 study record에서 직접 확인합니다.

```python
# 기전 → 마우스 뇌 발현 데이터에서 자주 살펴보는 표지 유전자(예시)
marker_genes = {
    "신경염증": ["Gfap", "Aif1", "Tnf", "Il1b"],
    "미토콘드리아 기능": ["Ppargc1a", "Pink1", "Prkn"],
    "산화스트레스": ["Sod1", "Sod2", "Nfe2l2", "Hmox1"],
    "단백질 응집": ["Snca", "Mapt", "App"],
    "시냅스 기능": ["Syp", "Dlg4", "Snap25"],
}

print("| 기전 | 표지 유전자(예) | 개수 |")
print("|---|---|---|")
for mechanism, genes in marker_genes.items():
    print(f"| {mechanism} | {', '.join(genes)} | {len(genes)} |")
```
