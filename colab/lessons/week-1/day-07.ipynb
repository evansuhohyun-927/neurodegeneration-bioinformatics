### 예시 2 — 교란변수가 만들어 낸 가짜 차이 (시뮬레이션)

가상의 세계를 하나 만듭니다. 이 세계의 규칙은 "보행 변동성은 **나이에만** 영향을 받고, 진단군과는 아무 관련이 없다"입니다. 다만 PD군이 평균 12살 더 많습니다.

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(7)
n = 200

# 가상 세계의 규칙: 보행 변동성은 "나이"에만 영향을 받고, 진단군과는 무관하다
group = np.array(["Control"] * n + ["PD"] * n)
age = np.concatenate([rng.normal(58, 6, n), rng.normal(70, 6, n)])   # PD군이 평균 12살 많다
stride_cv = 1.0 + 0.05 * (age - 60) + rng.normal(0, 0.3, 2 * n)

df = pd.DataFrame({"group": group, "age": age, "stride_cv": stride_cv})

print("① 군별 평균 (나이를 무시했을 때)")
print(df.groupby("group")["stride_cv"].mean().round(2))

df["age_band"] = pd.cut(df["age"], bins=[0, 60, 65, 70, 120], labels=["~60", "60-65", "65-70", "70~"])
print()
print("② 비슷한 나이끼리 묶어서 본 군별 평균")
print(df.pivot_table(index="age_band", columns="group", values="stride_cv", aggfunc="mean", observed=True).round(2))
```
