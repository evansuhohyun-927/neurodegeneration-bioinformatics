## 개념 설명

### 조건문은 프로토콜의 "만약 ~이면"이다

실험 프로토콜에는 이런 문장이 흔합니다. "RNA 농도가 50 ng/µL 이상이면 다음 단계로, 아니면 다시 추출." 상황에 따라 할 일이 달라지는 것이죠. 코드에서 이 역할을 하는 것이 조건문(conditional)입니다.

```python
expression = 12.5   # 예: 어떤 유전자의 발현값
if expression > 10:
    print("발현이 높다")
else:
    print("발현이 높지 않다")
```

### 반복문은 멀티채널 피펫이다

96-well 플레이트의 well마다 같은 시약을 넣어야 한다면, 한 칸씩 96번 반복하는 대신 멀티채널 피펫을 씁니다. 같은 동작을 여러 대상에 되풀이하는 것, 이것이 반복문(loop)입니다.

```python
genes = ["Gfap", "Aif1", "Tnf"]   # 예: 유전자 이름 리스트
for gene in genes:
    print(gene)
```

## 예시로 이해하기

### 예시 1 — if / elif / else

```python
expression = 4.0

if expression > 10:
    label = "high"
elif expression > 3:
    label = "medium"
else:
    label = "low"

print(label)
```

### 예시 2 — for로 하나씩 꺼내기

```python
genes = ["Gfap", "Aif1", "Tnf"]

for gene in genes:
    print(f"지금 보는 유전자: {gene}")

print("반복 끝")
```

### 예시 3 — 임계값을 넘는 유전자만 고르기

```python
expression = {
    "Gfap": 12.5,
    "Aif1": 8.9,
    "Tnf": 2.1,
    "Il1b": 1.4,
    "Sod2": 15.1,
}
threshold = 5

high_genes = []

for gene, value in expression.items():
    if value > threshold:
        high_genes.append(gene)

print(high_genes)
print(f"임계값 {threshold} 초과 유전자 수: {len(high_genes)}")
```

### 예시 4 — 조건별 샘플 수 세기

데이터를 받으면 가장 먼저 "각 군에 샘플이 몇 개인가"를 확인합니다. 그 일을 손으로 짜 봅니다.

```python
sample_condition = {
    "S1": "flight",
    "S2": "flight",
    "S3": "ground",
    "S4": "ground",
    "S5": "flight",
}

n_flight = 0
n_ground = 0

for sample, condition in sample_condition.items():
    if condition == "flight":
        n_flight = n_flight + 1
    elif condition == "ground":
        n_ground = n_ground + 1

print(f"flight: {n_flight}개, ground: {n_ground}개")
```

### 예시 5 — 비율로 방향 판정하기

```python
flight = {"Gfap": 250.0, "Tnf": 30.0, "Sod2": 80.0}
ground = {"Gfap": 100.0, "Tnf": 28.0, "Sod2": 160.0}

for gene in flight:
    ratio = flight[gene] / ground[gene]
    if ratio >= 2:
        direction = "증가"
    elif ratio <= 0.5:
        direction = "감소"
    else:
        direction = "큰 변화 없음"
    print(f"{gene}: 비율 {round(ratio, 2)} → {direction}")
```

4. **숫자 반복.** 아래를 실행해 `range(3)`이 0, 1, 2를 차례로 내놓는 것을 확인하고, 리스트의 합과 평균을 반복문으로 직접 구합니다.

```python
for i in range(3):
    print(i)

values = [12.5, 3.2, 8.9]
total = 0
for v in values:
    total = total + v
print(total)
print(round(total / len(values), 2))
```
