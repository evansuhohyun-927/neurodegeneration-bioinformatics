## 예시로 이해하기

### 예시 1 — 네 가지 자료형 만들어 보기

```python
gene = "Gfap"
expression = 12.5
n_samples = 6
is_control = True

print(gene)
print(expression)
print(n_samples)
print(is_control)
print(type(gene))
print(type(expression))
print(type(n_samples))
print(type(is_control))
```

### 예시 2 — 숫자 계산

우주비행군(flight)과 지상 대조군(ground)의 가상 발현값으로 차이와 비율을 구합니다.

```python
flight = 250.0
ground = 100.0

difference = flight - ground
ratio = flight / ground

print(difference)
print(ratio)
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(2 ** 10)
```

### 예시 3 — 문자열 다루기와 f-string

```python
gene = "gfap"
print(gene.upper())
print(gene.capitalize())
print(len(gene))
print(gene + "_mouse")

expression = 12.5
print(f"{gene} 유전자의 발현값은 {expression}입니다.")
```

### 예시 4 — 비교하면 bool이 나온다

```python
expression = 12.5
threshold = 10

print(expression > threshold)
print(expression == threshold)
print(expression != threshold)

is_high = expression > threshold
print(is_high)
print(type(is_high))
```

### 예시 5 — 글자 "12.5"를 숫자 12.5로

```python
value_text = "12.5"
print(type(value_text))

value_number = float(value_text)
print(value_number + 1)
print(int("6") * 2)
print(str(12.5) + " TPM")
```
